```python
from pathlib import Path
import json

json_path = (
    Path.cwd().parent
    / "storage"
    / "processed"
    / "chunks_with_embeddings.json"
)

with open(json_path) as f:
    documents = json.load(f)

print(len(documents))
print(documents[0].keys())
```

```python
print(documents[0].keys())
```

```python
from tqdm import tqdm

for doc in tqdm(documents):

    es.index(
        index="pdf_documents",
        id=doc["id"],          # keeps document IDs stable
        document={
            "source": doc["source"],
            "chunk_id": doc["chunk_id"],
            "text": doc["text"],
            "embedding": doc["embedding"],
        }
    )

print("Done!")
```

```python
es.count(index="pdf_documents")
```

```python
response = es.search(
    index="pdf_documents",
    query={
        "match": {
            "text": "dark matter"
        }
    },
    size=5
)

for hit in response["hits"]["hits"]:
    print("=" * 80)
    print(hit["_score"])
    print(hit["_source"]["source"])
    print(hit["_source"]["text"][:300])
```
